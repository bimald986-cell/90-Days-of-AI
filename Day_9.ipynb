```python
import pandas as pd

data = {
    "item": ["Rice", "Oil", "Soap", "Milk", "Tea"],
    "category": ["Food", "Food", "Home", "Food", "Drink"],
    "price": [1200, 350, 100, 180, 250],
    "stock": [20, 8, 40, 5, 15]
}

df = pd.DataFrame(data)

print(df)
```

```python
low_stock = df[df["stock"] < 10]

print(low_stock)
```

```python
low_stock = df[df["stock"] < 10]

print(low_stock)
print("Number of low-stock items:", len(low_stock))
```

```python
urgent_food = df[
    (df["category"] == "Food") &
    (df["stock"] < 10)
]

print(urgent_food)
```

```python
result = df[
    (df["category"] == "Food") |
    (df["stock"] < 10)
]

print(result)
```

```python
result = df[
    (df["price"] > 300) |
    (df["stock"] < 10)
]

print(result)
```

```python
df["stock_value"] = df["price"] * df["stock"]

print(df)
```

```python
total_stock_value = df["stock_value"].sum()

print("Total stock value:", total_stock_value)
```

```python
print("Total stock value:", df["stock_value"].sum())
print("Average price:", df["price"].mean())
print("Highest price:", df["price"].max())
print("Lowest price:", df["price"].min())
```

```python
print("SHOP INVENTORY SUMMARY")
print("----------------------")

print("Number of products:", len(df))
print("Total stock units:", df["stock"].sum())
print("Total stock value:", df["stock_value"].sum())
print("Average product price:", df["price"].mean())
print("Highest product price:", df["price"].max())
print("Lowest product price:", df["price"].min())
```

```python
import matplotlib.pyplot as plt

plt.bar(df["item"], df["stock_value"])

plt.title("Stock Value by Product")
plt.xlabel("Product")
plt.ylabel("Stock Value")

plt.show()
```

```python
sorted_stock = df.sort_values("stock_value", ascending=False)

print(sorted_stock)
```
