```python
scores = np.array([72, 85, 91, 68, 88, 95])
```

```python
ratings = np.where(
    scores >= 90,
    "Excellent",
    np.where(
        scores >= 80,
        "Good",
        "Review"
    )
)

ratings
```

```python
decision = np.where(
    scores >= 90,
    "Hire",
    np.where(
        scores >= 75,
        "Interview",
        "Reject"
    )
)

decision
```

```python
import pandas as pd

candidates = pd.DataFrame({
    "Name": ["Asha", "Ravi", "Maya", "Omar", "Sara", "Daniel"],
    "Score": scores,
    "Decision": decision
})

candidates
```

```python
hired_candidates = candidates[candidates["Decision"] == "Hire"]

hired_candidates
```

```python
shortlisted_candidates = candidates[candidates["Decision"] == "Interview" | candidates["Decision"] == "Interview" & candidates["Score"] >= 85]

shortlisted_candidates
```

```python
shortlisted_candidates = candidates[(candidates["Decision"] == "Hire") | (candidates["Decision"] == "Interview") & (candidates["Score"] >= 85)]

shortlisted_candidates
```

```python
shortlisted_candidates = candidates[
    (
        (candidates["Decision"] == "Hire") |
        (candidates["Decision"] == "Interview")
    )
    & (candidates["Score"] >= 85)
]

shortlisted_candidates
```

```python
shortlisted_candidates = candidates[
    (candidates["Decision"] != "Reject") &
    (candidates["Score"] >= 85)
]

shortlisted_candidates
```

```python
ranked_shortlist = shortlisted_candidates.sort_values(by="Score", ascending=False)
ranked_shortlist
```
