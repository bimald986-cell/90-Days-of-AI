```python
ideas = []
num_ideas = int(input("How many ideas do you want to create?"))
for i in range(num_ideas):
  idea = input(f"Enter idea {i+1}: ")
  ideas.append(idea)
  print("\nIs time real?:\n")
  for idea in ideas:
    print("-", idea)

print("\nTotal ideas:", len(ideas))
```
