```python
name = input("What is your name? ")
age = int(input("How old are you? "))
learning_hours = float(input("How many hours can you study today? "))

if learning_hours >= 2:
  plan = "Deep learning session"
elif learning_hours >= 1:
  plan = "Standard learning session"
elif learning_hours == 0:
  plan = "Quick learning session"
else:
  plan = "Rest today, but come back tomorrow."

print("\nHello", name)
print("Age:", age)
print("Today's plan:", plan)
```
