```python
import pandas as pd
import numpy as np

candidates = {
    "Name": ["Asha", "Ravi", "Maya", "Omar", "Sara", "Daniel"],
    "Subject": ["Math", "English", "Science", "Math", "English", "Science"],
    "Experience": [5, 2, 7, 3, 6, 4],
    "ExpectedSalary": [1200, 900, 1500, 1100, 1300, 1250]
}

df = pd.DataFrame(candidates)

df
```

```python
df[df["Experience"] >= 5]
```

```python
df["SeniorCandidate"] = df["Experience"] >= 5

df
```

```python
df["HighSalary"] = df["ExpectedSalary"] >= 1300
df
```

```python
df["PriorityCandidate"] = (df["Experience"] >= 5) & (df["ExpectedSalary"] <= 1300)
df
```

```python
df["SalaryCategory"] = np.where(
    df["ExpectedSalary"] >= 1300,
    "High",
    "Standard"
)
df
```

```python
conditions = [df["Experience"] >= 5, df["Experience"] >= 4]
choices = ["Senior", "Mid"]
df["ExperienceLevel"] = np.select(
    conditions,
    choices,
    default="Junior"
)
df
```

```python
conditions = [
    df["Experience"] >= 5,
    df["Experience"] >= 3
]

choices = [
    "Senior",
    "Mid"
]

df["ExperienceLevel"] = np.select(
    conditions,
    choices,
    default="Junior"
)

df
```

```python
df["MatchScore"] = (
    (df["Experience"] >= 5) * 2
    + (df["ExpectedSalary"] <= 1300) * 1
)

df
```

```python
df["OPENScore"] = (
    (df["Experience"] >= 5) * 3
    + (df["ExpectedSalary"] <= 1300) * 2 + (df["Subject"] == "Science") * 1
)

df
```

```python
df["MatchDecision"] = np.where(
    df["OPENScore"] >= 4,
    "Strong Match",
    "review"
)
df
```

```python
df["MatchDecision"] = np.where(
    df["OPENScore"] >= 4,
    "Strong Match",
    "Review"
)
df
```

```python
df["ranked_candidates"] = filtered = df["MatchDecision"]

sorted_data = filtered.sort_values(
    "OPENScore",
    ascending=False
)

result = sorted_data[
    ["Name", "Subject", "Experience", "ExpectedSalary", "OPENScore"]
]

result
df["ranked_candidates"]
```
