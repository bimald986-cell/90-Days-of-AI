```python
Employee = [
    {
        "name": "Ram",
        "department": "Finance",
        "years_service": 10,
        "salary": 1500
    },
    {
        "name": "Shyam",
        "department": "IT",
        "years_service": 7,
        "salary": 1350
    },
    {
        "name": "Hari",
        "department": "Security",
        "years_service": 12,
        "salary": 950
    },
    {
        "name": "John",
        "department": "HR",
        "years_service": 3,
        "salary": 800
    }
]

print(Employee[2]["name"])
```

```python
for employee in Employee:
    if employee["years_service"] >= 5:
        print(employee["name"], "Eligible for Long Service Recognition")
    else:
        print(employee["name"], "Not yet eligible")
```

```python
for employee in Employee:
    if employee["years_service"] >= 10:
        print(employee["name"], "Gold Recognition")

    elif 5 <= employee["years_service"] <= 9:
            print(employee["name"], "Silver Recognition")
    else:
        print(employee["name"], "Not yet eligible")
```
