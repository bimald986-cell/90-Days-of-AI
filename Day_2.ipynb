```python
employee = {
    "name": "John Smith",
    "department": "IT",
    "salary": 650,
    "years_service": 6
}

print(employee)
```

```python
print(employee["name"])
print(employee["salary"])
```

```python
if employee["years_service"] >= 5:
    print(employee["name"], "is a long-service employee.")
else:
    print(employee["name"], "has less than 5 years of service.")
```

```python
employees = [
    {
        "name": "Ahmed",
        "department": "HR",
        "salary": 600
    },
    {
        "name": "Sara",
        "department": "Finance",
        "salary": 750
    },
    {
        "name": "David",
        "department": "IT",
        "salary": 900
    }
]

print(employees)
```

```python
print(employees[1]["name"])
```

```python
for employee in employees:
    print(employee["name"], "-", employee["department"])
```
